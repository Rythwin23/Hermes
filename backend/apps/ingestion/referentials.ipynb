```python
import polars as pl
```

## List of stops

```python
stops = pl.scan_csv("../../ressources/stops.txt")
stops.collect()
```

```python
stops.filter(pl.col("stop_id") == "IDFM:monomodalStopPlace:411440").collect()
```

```python
stops.filter(pl.col("stop_name").str.ends_with("Les Halles")).select("stop_id", "stop_name", "parent_station").collect()
```

## Parents stations

```python
parent_stops = (
    stops.lazy()
    .filter(pl.col("parent_station").is_null())
)
parent_stops.collect()
```

## Child stations

```python
child_stops = (
    stops.lazy()
    .filter(pl.col("parent_station").is_not_null())
)
child_stops.collect()
```

## Parent station with all child stations

```python
parent_child_stops = (
    parent_stops.lazy()
    .select(["stop_id", "stop_name"])
    .join(
        child_stops.lazy(),
        left_on="stop_id",
        right_on="parent_station",
        how="left")
    .group_by(["stop_id", "stop_name"])
    .agg(pl.col("stop_id_right").alias("child_stations"),
         pl.col("stop_name_right").alias("child_stations_names")
    )
)
parent_child_stops.collect()
```

## RER - metro routes

```python
routes = pl.scan_csv("../../ressources/routes.txt")

rer_metro_routes = (
    routes.lazy()
    .filter(pl.col("route_type").is_in([1, 2]))
    .sort("route_short_name")
)
rer_metro_routes.collect()
```

## Trips

```python
trips = pl.scan_csv("../../ressources/trips.txt")
trips.collect()
```

## Trips of metro and RER

```python
rer_metro_trips = (
    rer_metro_routes.lazy()
    .select("route_id", "route_long_name")
    .join(trips.lazy(),
          on="route_id",
          how="inner")
    .sort("route_long_name")
)
rer_metro_trips.collect()
```

## Exemple for RER D trips

```python
stop_times = pl.scan_csv("../../ressources/stop_times.txt")
stop_times.collect()
```

## Trips stop times

```python
# trip rer A

rer_metro_trips_stop_times = (
    rer_metro_trips.lazy()
    .select(["trip_id", "route_id", "route_long_name", "service_id"])
    .join(stop_times.lazy(),
          on="trip_id",
          how="inner")
    .sort("route_long_name")
)
rer_metro_trips_stop_times.collect()
```

```python
def gtfs_time_to_pl(col: str) -> pl.Expr:
    return pl.time(
        hour=pl.col(col).str.extract(r"^(\d+):\d{2}:\d{2}$", 1).cast(pl.Int64) % 24,
        minute=pl.col(col).str.extract(r"^\d+:(\d{2}):\d{2}$", 1).cast(pl.Int64),
        second=pl.col(col).str.extract(r"^\d+:\d{2}:(\d{2})$", 1).cast(pl.Int64),
    ).alias(col)

rer_metro_trips_stop_times = (
    rer_metro_trips_stop_times
    .with_columns(
        gtfs_time_to_pl("departure_time"),
        gtfs_time_to_pl("arrival_time"),
    )
)
rer_metro_trips_stop_times.collect()
```

## EXEMPLE LIGNE 1 METRO avec stop name

```python
stops.join(
    rer_metro_trips_stop_times
        .filter(pl.col("route_id")=="IDFM:C01371")
        .select(["trip_id", "service_id", "stop_id", "route_id", "route_long_name" , "arrival_time", "departure_time"]),
    on="stop_id",
    how="inner"
).select(["trip_id", "service_id", "stop_id", "stop_name","route_id", "route_long_name", "arrival_time", "departure_time"]).sort("arrival_time").collect()
```

## RER Route by Parent_stop

```python
rer_metro_trips_stops = (
    rer_metro_trips_stop_times.lazy()
    .group_by(["trip_id", "route_id", "route_long_name", "service_id"])
    .agg(pl.col("stop_id")
         .alias("stops"))
)
rer_metro_trips_stops.collect()
```

```python
parent_stops_routes = (
    rer_metro_trips_stops.lazy()
    .explode("stops", empty_as_null=True)
    .rename({"stops": "stop_id"})
    # Récupérer le parent_station de chaque stop
    .join(
        stops.lazy().select(["stop_id", "parent_station"]),
        on="stop_id",
        how="left"
    )
    # Si pas de parent, utiliser le stop lui-même
    .with_columns(
        pl.coalesce(["parent_station", "stop_id"]).alias("parent_station")
    )
    # Récupérer le nom du parent
    .join(
        stops.lazy().select(["stop_id", "stop_name"]),
        left_on="parent_station",
        right_on="stop_id",
        how="left"
    )
    # Grouper par station parente
    .group_by(["parent_station", "stop_name"])
    .agg(
        pl.col("route_long_name").unique().alias("routes"),
        pl.col("route_id").unique().alias("routes_ids")
    )
    .sort("stop_name")
)

parent_stops_routes.collect()
```

```python
gares = "gare de l'est|gare du nord|châtelet|halles|gare de lyon|austerlitz|lazare"
exp = parent_stops_routes.filter(pl.col("stop_name").str.to_lowercase().str.contains(gares).alias("regex")).select(["parent_station", "stop_name", "routes"]).collect()
```
