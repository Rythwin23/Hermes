# Polars-based parsing data existing in ressources folder

```python
import polars as pl
```

```python
df_stops = pl.read_csv("../ressources/stops.txt")
df_stops.head()
```

# routes.txt description
- route_type :
    - 0 : tram
    - 1 : metro
    - 2 : RER TER
    - 3 : bus       

```python
df_routes = pl.read_csv("../ressources/routes.txt")
df_routes.head()
```

```python
df_routes.filter(pl.col("route_type").is_in([1,2])).select(["route_id", "route_long_name", "route_type"]).sort("route_id")
```

```python
df_trips = pl.read_csv("../ressources/trips.txt")
df_trips.head()
```

```python
# exemple of a specific line (exemple RER A)
df_trips.join(df_routes, on="route_id", how="inner").filter(pl.col("route_type").eq(2), pl.col("route_long_name").eq("A"))
```

```python
df_transfers = pl.read_csv("../ressources/transfers.txt")
df_transfers.head()
```

```python
# exemple of transfers between stations chatelet
stop_id_chatelet = df_stops.filter(pl.col("stop_name").eq("Châtelet - Les Halles")).select("stop_id").to_series()[0]
df_transfers.filter(pl.col("from_stop_id").eq(stop_id_chatelet)).join(df_stops, left_on="to_stop_id", right_on="stop_id", how="inner").select(["transfer_type", "stop_name"]).unique()
```
