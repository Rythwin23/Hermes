```python
import polars as pl
```

## List of stops

```python
stops = pl.scan_csv("../ressources/stops.txt")
stops.collect()
```

## Parents stations

```python
parent_stops = (
    stops.lazy()
    .filter(pl.col("parent_station").is_null())
)
parent_stops.collect()
```

## Child stations

```python
child_stops = (
    stops.lazy()
    .filter(pl.col("parent_station").is_not_null())
)
child_stops.collect()
```

## Parent station with all child stations

```python
parent_child_stops = (
    parent_stops.lazy()
    .select(["stop_id", "stop_name"])
    .join(
        child_stops.lazy(),
        left_on="stop_id",
        right_on="parent_station",
        how="left")
    .group_by(["stop_id", "stop_name"])
    .agg(pl.col("stop_id_right").alias("child_stations"),
         pl.col("stop_name_right").alias("child_stations_names")
    )
)
parent_child_stops.collect()
```

## RER routes

```python
routes = pl.scan_csv("../ressources/routes.txt")

rer_routes = (
    routes.lazy()
    .filter(pl.col("route_type") == 2)
    .sort("route_short_name")
)
rer_routes.collect()
```

## RER trips

```python
trips = pl.scan_csv("../ressources/trips.txt")

rer_trips = (
    rer_routes.lazy()
    .select("route_id", "route_long_name")
    .join(trips.lazy(),
          on="route_id",
          how="inner")
    .sort("route_long_name")
)
rer_trips.collect()
```

## Trips stop times

```python
# trip rer A
stop_times = pl.scan_csv("../ressources/stop_times.txt")
rer_trips_stop_times = (
    rer_trips.lazy()
    .select(["trip_id", "route_id", "route_long_name", "service_id"])
    .join(stop_times.lazy(),
          on="trip_id",
          how="inner")
    .sort("route_long_name")
)
rer_trips_stop_times.collect()
```

## RER Route by Parent_stop

```python
rer_trips_stops = (
    rer_trips_stop_times.lazy()
    .group_by(["trip_id", "route_id", "route_long_name", "service_id"])
    .agg(pl.col("stop_id")
         .alias("stops"))
)
rer_trips_stops.collect()
```

```python
parent_stops_routes = (
    rer_trips_stops.lazy()
    .explode("stops", empty_as_null=True)
    .rename({"stops": "stop_id"})
    .join(
        stops.lazy()
        .select(["stop_id", "stop_name"]),
        on="stop_id",
        how="left"
    )
    .group_by(["stop_id", "stop_name"])
    .agg(
        pl.col("route_long_name")
        .unique()
        .alias("routes"),
        pl.col("route_id")
        .unique()
        .alias("routes_ids")
    )
    .sort("stop_name")
)

parent_stops_routes.collect()
```
